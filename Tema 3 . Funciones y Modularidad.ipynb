```python
# funcion
# * args, Recibe todos los argumentos posicionales en una tupla.
# **Kwargs, Recibe todos los argumentos con nombre en un diccionario
def myFun(*args,**kwargs):
  print(args)
  print(kwargs)
myFun('abc', 'def', x=1, y=2, z=3)
```

```python
# Función anidada
def f1():
    s = "Hola mundo"
    def f2():
        print(s)
    f2()
#--------------
f1()
```

```python
def myFun(x):
    def myInnerFun(y):
        return x + y
    return myInnerFun

myFun(2)(3)
```
