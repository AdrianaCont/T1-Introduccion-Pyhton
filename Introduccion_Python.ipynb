## 1. Variables

Una variable es un nombre que guarda un valor. En Python **no hace falta declarar el tipo**: se asigna con `=`.

````python
nombre = "Laura"         # str (texto)
edad = 28               # int (entero)
estatura = 1.65         # float (decimal)
es_estudiante = False   # bool (verdadero/falso)

#Esto es un comentario

print(nombre)
print(edad)
print(estatura)
print(es_estudiante)
````

Con `type()` puedes ver el tipo de dato de una variable:

````python
print(type(nombre))
print(type(edad))
print(type(estatura))
print(type(es_estudiante))
````

Puedes cambiar el valor (y hasta el tipo) de una variable, y asignar varias a la vez:

````python
edad = 26
print("Nueva edad:", edad)

# Asignación múltiple
x, y, z = 1, 2, 3
print("variables", x, y, z)

# Mismo valor para varias variables
a = b = c = 10
print(a,b,c)
# Imprimir texto con variables (f-string)
print(f"Me llamo {nombre} y tengo {edad} años")
````

**Reglas para nombrar variables:**
- Pueden contener letras, números y guion bajo `_`
- No pueden empezar con número
- No pueden tener espacios ni ser palabras reservadas (`if`, `for`, `while`...)
- Python distingue mayúsculas de minúsculas: `edad` ≠ `Edad`

## 2. Operadores

### 2.1 Operadores aritméticos

````python
a = 1
b = 6

print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicación:", a * b)
print("División:", a / b)             # siempre devuelve float
print("División entera:", a // b)     # descarta los decimales
print("Módulo (residuo):", a % b)     # modulo
print("Potencia:", a ** b)            # 10 elevado a la 3
````

### 2.2 Operadores de asignación

````python
n = 10        # asignación simple

n += 5        # equivale a n = n + 5
print("n += 5  ->", n)

n -= 3        # n = n - 3
print("n -= 3  ->", n)

n *= 2        # n = n * 2
print("n *= 2  ->", n)

n /= 4        # n = n / 4
print("n /= 4  ->", n)

n = 17
n //= 5       # n = n // 5
print("n //= 5 ->", n)

n = 17
n %= 5        # n = n % 5
print("n %= 5  ->", n)

n = 2
n **= 3       # n = n ** 3
print("n **= 3 ->", n)
````

### 2.3 Operadores de comparación (relacionales)

Comparan dos valores y devuelven `True` o `False`.

````python
a = 7
b = 10

print("a == b :", a == b)   # igual a
print("a != b :", a != b)   # distinto de
print("a >  b :", a > b)    # mayor que
print("a <  b :", a < b)    # menor que
print("a >= b :", a >= b)   # mayor o igual que
print("a <= b :", a <= b)   # menor o igual que
````

**Cuidado:** `=` *asigna* un valor y `==` *compara* dos valores.

````python
print("hola" == "hola")
print("hola" == "Hola")   # distingue mayúsculas
print(5 == 5.0)           # True: mismo valor numérico
````

### 2.4 Operadores lógicos

Combinan condiciones: `and` (y), `or` (o), `not` (negación).

````python
edad = 20
tiene_licencia = True

print(edad >= 18 and tiene_licencia)   # True solo si AMBAS son True
print(edad < 18 or tiene_licencia)     # True si AL MENOS UNA es True
print(not tiene_licencia)              # invierte el valor
````

**Tabla de verdad rápida**

### Tablas de verdad de los operadores lógicos

**`and` (y):** es `True` solo si ambas son `True`.

| p | q | p and q |
|:-----:|:-----:|:-------:|
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |

**`or` (o):** es `True` si al menos una es `True`.

| p | q | p or q |
|:-----:|:-----:|:------:|
| True | True | True |
| True | False | True |
| False | True | True |
| False | False | False |

**`not` (negación):** invierte el valor.

| p | not p |
|:-----:|:-----:|
| True | False |
| False | True |

**Todas juntas:**

| p | q | p and q | p or q | not p |
|:-----:|:-----:|:-------:|:------:|:-----:|
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

### 2.5 Operador ternario

Es una forma **compacta** de escribir un `if / else` en una sola línea:

```
valor_si_verdadero if condicion else valor_si_falso
```

````python
edad = 17
estado = "mayor de edad" if edad >= 18 else "menor de edad"
print(estado)

numero = 3
tipo = "par" if numero % 2 == 0 else "impar"
print(f"El {numero} es {tipo}")

# Otro ejemplo: el mayor de dos números
a, b = 4, 9
mayor = a if a > b else b
print("El mayor es:", mayor)
````
