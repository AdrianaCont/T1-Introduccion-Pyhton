```python
# Instrucciones: En una celda de código, define las siguientes variables para un solo registro:

id_registro = 101 # Número entero
lectura_valor = 23.5 # Número decimal (flotante)
nombre_sensor = "Sensor_Temp_01" # Cadena de texto
es_valido = True # Valor booleano inicializado en True
```

```python
# Instrucciones:

# 1. Crea una variable factor_ajuste con valor 1.05
factor_ajuste = 1.05
# 2. Calcula valor_normalizado multiplicando lectura original por factor y restando constante de error (0.2)
valor_normalizado = (lectura_valor * factor_ajuste) - 0.2
# 3. Calcula el cuadrado del error usando el operador de potencia (**)
error_cuadratico = (lectura_valor - valor_normalizado) ** 2
```

```python
# Instrucciones:

# 1. Define un limite_superior de 100.0
limite_superior = 100.0

# 2. Actualiza la variable es_valido comparando si valor_normalizado es menor al límite
es_valido = valor_normalizado < limite_superior

# 3. Imprime un reporte final usando f-strings
print(f"Sensor: {nombre_sensor} | Valor Final: {valor_normalizado:.2f} | Válido: {es_valido}")
```

```python
# Instrucciones:

# 1. Encapsula el código en una función
def procesar_lectura(valor):
    """Procesa y valida lecturas del sensor de temperatura.
    Args:
        valor (float): Lectura original cruda del sensor.
    Returns:
        tuple: (valor_final, estatus_booleano) con valor transformado y validación.
    """
    factor_ajuste = 1.05 # Tipo float por precisión decimal
    valor_final = (valor * factor_ajuste) - 0.2
    estatus_booleano = valor_final < 100.0
    return (valor_final, estatus_booleano)
```
